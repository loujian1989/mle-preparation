```python
# TTL = "this entry is only valid for N seconds after I store it." ML analogy: "last clicked item" features go stale after 10 minutes.
# ttl_seconds is the TTL we set

import time 
cache = {}

def set(key, value, ttl_seconds):
    cache[key] = (value, time.time() + ttl_seconds)

def get(key):
    if key not in cache:
        return None
    value, expiry = cache[key]
    if time.time() > expiry:
        del cache[key]
        return None
    return cache[key]
    
```
